Exercise 3 A mini nutri-scan database

```python
nutrition = {
    "banana": {
        "calories": 89,
        "protein": 1.1
    },

    "egg": {
        "calories": 155,
        "protein": 13
    }
}
print(nutrition["banana"]["calories"])
print(nutrition["banana"]["protein"])
print(nutrition["egg"]["calories"])
print(nutrition["egg"] ["protein"])
```
